Small example showing how to load the models and checkpoints from the package.

```python
from pathlib import Path
import torch
from cbam_kd.checkpoint import load_model_checkpoint
from cbam_kd.models.resnet_cifar import resnet20, resnet56

ROOT = Path("..").resolve()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
teacher = resnet56(100).to(device)
baseline = resnet20(100).to(device)
kd_student = resnet20(100).to(device)

load_model_checkpoint(teacher, ROOT / "checkpoints/resnet56_cifar100_best.pth", map_location=device)
load_model_checkpoint(baseline, ROOT / "checkpoints/resnet20_cifar100_best.pth", map_location=device)
load_model_checkpoint(kd_student, ROOT / "checkpoints/resnet20_kd_best.pth", map_location=device)
print("All historical checkpoints loaded exactly.")
```

```python
images = torch.randn(4, 3, 32, 32, device=device)
with torch.no_grad():
    logits, features = kd_student.forward_with_features(images)
print(logits.shape)
print([tuple(feature.shape) for feature in features])
```

Training and evaluation are run from the scripts described in the README.
